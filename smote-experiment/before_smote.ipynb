```python
import os
import pandas as pd

from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix, f1_score, precision_score, recall_score
```

```python
RANDOM_STATE = 42

base_dir = "/content/drive/MyDrive/fraud processed data"
train_path = os.path.join(base_dir, "train_processed.csv")
test_path = os.path.join(base_dir, "test_processed.csv")

outdir = os.path.join(base_dir, "results_before_smote")
os.makedirs(outdir, exist_ok=True)

print("Train path:", train_path)
print("Test path :", test_path)
print("Output dir:", outdir)
```

```python
train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

print("Train shape:", train_df.shape)
print("Test shape :", test_df.shape)

print("\nTrain columns:")
print(train_df.columns.tolist())
```

```python
X_train = train_df.drop(columns=["isFraud"])
y_train = train_df["isFraud"]

X_test = test_df.drop(columns=["isFraud"])
y_test = test_df["isFraud"]

print("X_train:", X_train.shape)
print("X_test :", X_test.shape)

print("\ny_train distribution:")
print(y_train.value_counts())

print("\ny_test distribution:")
print(y_test.value_counts())
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, classification_report

print("Training RandomForest baseline (before SMOTE)...")

model = RandomForestClassifier(
     n_estimators=15,
    random_state=RANDOM_STATE,
    n_jobs=-1,
)

model.fit(X_train, y_train)

# Predictions
y_train_pred = model.predict(X_train)
y_test_pred = model.predict(X_test)

# Accuracy
train_acc = accuracy_score(y_train, y_train_pred)
test_acc = accuracy_score(y_test, y_test_pred)

# Fraud metrics on test
fraud_precision = precision_score(y_test, y_test_pred, pos_label=1, zero_division=0)
fraud_recall = recall_score(y_test, y_test_pred, pos_label=1, zero_division=0)
fraud_f1 = f1_score(y_test, y_test_pred, pos_label=1, zero_division=0)
cm = confusion_matrix(y_test, y_test_pred)
report_text = classification_report(y_test, y_test_pred, zero_division=0, digits=4)

print("=" * 80)
print("RandomForest BEFORE SMOTE")
print("=" * 80)
print("Accuracy train:", round(train_acc, 6))
print("Accuracy test :", round(test_acc, 6))
print("Precision (fraud):", round(fraud_precision, 4))
print("Recall    (fraud):", round(fraud_recall, 4))
print("F1        (fraud):", round(fraud_f1, 4))
print("Confusion matrix:")
print(cm)
print("\nClassification report:\n")
print(report_text)
```

```python
# get dummies
#  بدال 
# label encoder
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, recall_score, f1_score

print("Training RandomForest with get_dummies...")

# البيانات الحالية
X_train_gd = train_df.drop(columns=["isFraud"]).copy()
y_train_gd = train_df["isFraud"].copy()

X_test_gd = test_df.drop(columns=["isFraud"]).copy()
y_test_gd = test_df["isFraud"].copy()

# تحويل type إلى dummies
X_train_gd = pd.get_dummies(X_train_gd, columns=["type"], prefix="type", dtype=int)
X_test_gd = pd.get_dummies(X_test_gd, columns=["type"], prefix="type", dtype=int)

# توحيد الأعمدة بين train و test
X_test_gd = X_test_gd.reindex(columns=X_train_gd.columns, fill_value=0)

# نفس موديل 
model_gd = RandomForestClassifier(
    n_estimators=15,
    random_state=RANDOM_STATE,
    n_jobs=-1,
)

model_gd.fit(X_train_gd, y_train_gd)

# التنبؤ
y_train_pred_gd = model_gd.predict(X_train_gd)
y_test_pred_gd = model_gd.predict(X_test_gd)

# المقاييس
train_acc_gd = accuracy_score(y_train_gd, y_train_pred_gd)
test_acc_gd = accuracy_score(y_test_gd, y_test_pred_gd)
recall_gd = recall_score(y_test_gd, y_test_pred_gd, pos_label=1, zero_division=0)
f1_gd = f1_score(y_test_gd, y_test_pred_gd, pos_label=1, zero_division=0)

print("=" * 60)
print("GET_DUMMIES RESULTS")
print("=" * 60)
print("Accuracy train:", round(train_acc_gd, 6))
print("Accuracy test :", round(test_acc_gd, 6))
print("Recall        :", round(recall_gd, 4))
print("F1-score      :", round(f1_gd, 4))
```

```python
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score, precision_score, recall_score, f1_score

# تنبؤات الاختبار
y_test_pred_gd = model_gd.predict(X_test_gd)

# المقاييس
train_acc_gd = accuracy_score(y_train_gd, model_gd.predict(X_train_gd))
test_acc_gd = accuracy_score(y_test_gd, y_test_pred_gd)
precision_gd = precision_score(y_test_gd, y_test_pred_gd, pos_label=1, zero_division=0)
recall_gd = recall_score(y_test_gd, y_test_pred_gd, pos_label=1, zero_division=0)
f1_gd = f1_score(y_test_gd, y_test_pred_gd, pos_label=1, zero_division=0)
cm_gd = confusion_matrix(y_test_gd, y_test_pred_gd)

print("=" * 60)
print("GET_DUMMIES RESULTS")
print("=" * 60)
print("Accuracy train:", round(train_acc_gd, 6))
print("Accuracy test :", round(test_acc_gd, 6))
print("Precision     :", round(precision_gd, 4))
print("Recall        :", round(recall_gd, 4))
print("F1-score      :", round(f1_gd, 4))
print("Confusion Matrix:")
print(cm_gd)

tn, fp, fn, tp = cm_gd.ravel()
print("\nTN:", tn)
print("FP:", fp)
print("FN:", fn)
print("TP:", tp)

print("\nClassification Report:\n")
print(classification_report(y_test_gd, y_test_pred_gd, zero_division=0, digits=4))
```
